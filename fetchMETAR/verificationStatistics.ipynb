```python
import re

# Example multiline string, but in practice you might read from a file or other source:
metar_text = """\
KPDX 280253Z 15008KT 9SM -RA SCT018 BKN023 OVC039 10/07 A2974 RMK AO2 SLP071 P0004 60015 T01000072 58014
KPDX 280153Z 16007KT 8SM RA FEW035 OVC048 10/07 A2975 RMK AO2 SLP072 P0008 T01000072
KPDX 280053Z 20011G19KT 10SM -RA FEW017 BKN050 OVC080 11/08 A2978 RMK AO2 SLP085 P0003 T01060078
KPDX 272353Z 20009KT 8SM RA FEW020 BKN055 OVC080 11/08 A2978 RMK AO2 RAB29 SLP085 P0001 60002 T01060078 10117 20106 50001
KPDX 272321Z 21010G19KT 10SM FEW021 BKN045 OVC110 11/08 A2979 RMK AO2 T01110078
KPDX 272253Z 20009KT 10SM FEW018 BKN027 OVC120 11/08 A2978 RMK AO2 SLP085 T01110078
KPDX 271853Z 21012KT 10SM FEW015 BKN049 BKN060 12/08 A2974 RMK AO2 SLP069 T01170083
KPDX 271753Z 19011G19KT 10SM SCT012 BKN037 11/08 A2971 RMK AO2 RAB10E47 SLP059 P0000 60004 T01060078 10106 20072 51030
KPDX 271748Z 19010G20KT 10SM SCT012 BKN037 11/08 A2971 RMK AO2 RAB10E47 P0000
KPDX 271701Z 20013G20KT 9SM SCT012 BKN029 BKN037 10/07 A2967 RMK AO2 T01000072
KPDX 271653Z 20012G18KT 10SM BKN012 BKN029 10/07 A2966 RMK AO2 RAB22E49 SLP044 P0000 T01000072
"""

def parse_metar_line(line):
    """
    Given a line like:
      KPDX 280253Z 15008KT 9SM -RA SCT018 ... T01000072 ...
    parse out:
      - date (day of month)
      - time (UTC, from HHMM in DDHHMMZ)
      - wind direction, speed, gust (if any)
      - temperature in Celsius (from the 'Txxxxyyyy' group)
    Return a dict with these fields.
    """

    # Split line into tokens
    tokens = line.split()

    result = {
        'station': None,
        'day': None,
        'time_utc': None,
        'wind_dir': None,
        'wind_speed_kt': None,
        'wind_gust_kt': None,
        'temperature_c': None,
    }

    # 1) Station (e.g. KPDX)
    if len(tokens) > 0:
        result['station'] = tokens[0]

    # 2) Date/time group (e.g. 280253Z)
    if len(tokens) > 1:
        dt_match = re.match(r"^(\d{2})(\d{2})(\d{2})Z$", tokens[1])
        if dt_match:
            day_str = dt_match.group(1)  # '28'
            hh = dt_match.group(2)       # '02'
            mm = dt_match.group(3)       # '53'
            result['day'] = day_str
            result['time_utc'] = f"{hh}:{mm}Z"

    # 3) Wind group: something like 15008KT or 20011G19KT
    # We'll look for the first token that matches that pattern
    wind_pattern = re.compile(r"^(\d{3})(\d{2})(G(\d{1,2}))?KT$")
    for tok in tokens:
        m = wind_pattern.match(tok)
        if m:
            # Direction is group(1), speed is group(2), gust is group(4)
            try:
                result['wind_dir'] = int(m.group(1))
                result['wind_speed_kt'] = int(m.group(2))
                if m.group(4):
                    result['wind_gust_kt'] = int(m.group(4))
            except ValueError:
                pass
            break

    # 4) Temperature from the T-group (e.g. T01000072)
    #    Format is TsTTTsddd, where:
    #       T = literal 'T'
    #       s = sign bit for temperature (0 => +, 1 => -)
    #       TTT = temperature in tenths of °C
    #       s = sign bit for dewpoint (0 => +, 1 => -)
    #       ddd = dewpoint in tenths of °C
    #
    #    Example: T01000072
    #      'T'  -> literal T
    #      '0'  -> temp sign bit (0 => +)
    #      '100'-> temperature in tenths => 10.0°C
    #      '0'  -> dewpoint sign bit (0 => +)
    #      '072'-> dewpoint in tenths => 7.2°C

    t_pattern = re.compile(r"^T(\d)(\d{3})(\d)(\d{3})$")
    for tok in tokens:
        m = t_pattern.match(tok)
        if m:
            # Sign bit for temperature
            temp_sign_bit = int(m.group(1))
            # Temperature (in tenths of °C)
            temp_tenths = int(m.group(2))

            # Convert to actual Celsius with sign
            if temp_sign_bit == 1:
                temperature_c = -temp_tenths / 10.0
            else:
                temperature_c = temp_tenths / 10.0

            result['temperature_c'] = temperature_c
            break  # Stop after parsing the first T-group found

    return result


def main():
    # We split the multiline string into individual lines (each is one METAR)
    lines = metar_text.strip().splitlines()

    # Parse each line
    observations = []
    for line in lines:
        parsed = parse_metar_line(line.strip())
        observations.append(parsed)

    # Print results
    for i, obs in enumerate(observations, start=1):
        print(f"--- Obs #{i} ---")
        print(f"  Station: {obs['station']}")
        print(f"  Day: {obs['day']}")
        print(f"  Time (UTC): {obs['time_utc']}")
        print(f"  Wind Dir (°): {obs['wind_dir']}")
        print(f"  Wind Speed (kt): {obs['wind_speed_kt']}")
        if obs['wind_gust_kt']:
            print(f"  Wind Gust (kt): {obs['wind_gust_kt']}")
        print(f"  Temperature (C): {obs['temperature_c']}\n")


if __name__ == "__main__":
    main()
```

```python
import os
import glob
import time
from datetime import datetime, timedelta, timezone
from urllib.request import urlopen
from io import StringIO
import pandas as pd

SERVICE = "http://mesonet.agron.iastate.edu/cgi-bin/request/asos.py?"
MAX_ATTEMPTS = 6
METAR_DIR = "./metar_csv"  # directory where raw CSVs are stored (created by downloader cell)


def download_data(uri):
    attempt = 0
    backoff = 5
    while attempt < MAX_ATTEMPTS:
        try:
            raw = urlopen(uri, timeout=300).read().decode("utf-8")
            if raw and not raw.startswith("ERROR"):
                return raw
        except Exception as exc:
            print(f"download attempt {attempt+1} failed: {exc}")
        attempt += 1
        time.sleep(backoff)
        backoff *= 2
    return ""


def fetch_asos_for_station(station, start_dt, end_dt, out_dir=METAR_DIR):
    service = (
        SERVICE
        + "data=all&tz=Etc/UTC&format=comma&latlon=yes&"
        + start_dt.strftime("year1=%Y&month1=%m&day1=%d&hour1=%H&minute1=%M&")
        + end_dt.strftime("year2=%Y&month2=%m&day2=%d&hour2=%H&minute2=%M&")
        + f"station={station}"
    )
    print(f"Requesting {station}: {start_dt.isoformat()} -> {end_dt.isoformat()}")
    raw = download_data(service)
    if not raw:
        print(f"No data for {station}")
        return None
    df = pd.read_csv(StringIO(raw), comment="#")
    if df.empty:
        print(f"Parsed empty dataframe for {station}")
        return None
    os.makedirs(out_dir, exist_ok=True)
    fn = f"{station}_{start_dt.strftime('%Y%m%dT%H%M')}_{end_dt.strftime('%Y%m%dT%H%M')}.csv"
    outpath = os.path.join(out_dir, fn)
    df.to_csv(outpath, index=False)
    print(f"Wrote {outpath} ({len(df)} rows)")
    return df


def summarize_metar_daily(df, valid_col="valid"):
    """
    Given a raw METAR DataFrame (IEM/ASOS format), keep tmpf, p01i, sknt,
    coerce to numeric, group into days defined 06Z-06Z and compute:
      - daily max temperature (F)
      - daily min temperature (F)
      - daily total precipitation (in)
      - daily max sustained wind speed (kt)
    Missing values are ignored. p01i NaN treated as 0 for accumulation.
    """
    # Ensure datetime
    df = df.copy()
    df[valid_col] = pd.to_datetime(df[valid_col], errors="coerce")
    df = df.dropna(subset=[valid_col])
    # Normalize columns (common names from IEM output)
    # tmpf = temperature Fahrenheit, p01i = precip inches, sknt = sustained wind (kt)
    for col in ["tmpf", "p01i", "sknt"]:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")
        else:
            # create missing column full of NaN so aggregations still work
            df[col] = pd.Series(pd.NA, index=df.index)

    # define METAR "day" as the local date when shifting time by -6h (so 06Z..06Z maps to same date)
    df["metar_day"] = (df[valid_col] - pd.Timedelta(hours=6)).dt.date

    agg = df.groupby("metar_day").agg(
        max_temp_f=("tmpf", "max"),
        min_temp_f=("tmpf", "min"),
        total_precip_in=("p01i", lambda x: x.fillna(0).sum()),
        max_wind_kt=("sknt", "max"),
        samples=("tmpf", "count"),
    )

    # Reorder and clean results
    agg = agg.reset_index().rename(columns={"metar_day": "date"})
    # convert date column to datetime.date -> keep as ISO-like string for CSV
    agg["date"] = agg["date"].astype(str)
    return agg


if __name__ == "__main__":
    # --- user input ---
    raw_input_str = input("Enter one or more station codes (comma separated, e.g. KDEN,KSLC): ").strip()
    if not raw_input_str:
        raise SystemExit("No station provided, aborting.")
    stations = [s.strip().upper() for s in raw_input_str.split(",") if s.strip()]

    # last 14 days UTC
    end_dt = datetime.now(timezone.utc).replace(second=0, microsecond=0)
    start_dt = end_dt - timedelta(days=14)

    os.makedirs(METAR_DIR, exist_ok=True)

    for st in stations:
        # try to find existing CSV for station in METAR_DIR
        pattern = os.path.join(METAR_DIR, f"{st}_*.csv")
        files = sorted(glob.glob(pattern))
        if files:
            # prefer the most recent file
            raw_df = pd.read_csv(files[-1])
            print(f"Loaded {files[-1]} ({len(raw_df)} rows)")
        else:
            # download if not present
            raw_df = fetch_asos_for_station(st, start_dt, end_dt, out_dir=METAR_DIR)
            if raw_df is None:
                print(f"No data for {st}; skipping.")
                continue

        summary = summarize_metar_daily(raw_df, valid_col="valid")
        outfn = os.path.join(METAR_DIR, f"{st}_daily_summary_06z_{start_dt.strftime('%Y%m%d')}_to_{end_dt.strftime('%Y%m%d')}.csv")
        summary.to_csv(outfn, index=False)
        print(f"Wrote daily summary {outfn}")
        print(summary.to_string(index=False))
```
